```python
"""
project_summary.py

Print an overview of all modules, classes and functions
defined in the project.
"""

import inspect
import importlib
import pkgutil


# ---------------------------------------------------------
# Print one module
# ---------------------------------------------------------

def summarize_module(module_name):

    module = importlib.import_module(module_name)

    print()
    print("=" * 80)
    print(module_name)
    print("=" * 80)

    # -----------------------------------------------------
    # Stand-alone functions
    # -----------------------------------------------------

    functions = []

    for name, func in inspect.getmembers(module, inspect.isfunction):

        if func.__module__ == module_name:

            functions.append((name, func))

    if functions:

        print("\nFunctions")

        for name, func in functions:

            print(
                f"  {name}{inspect.signature(func)}"
            )

            doc = inspect.getdoc(func)

            if doc:
                print(f"      {doc.splitlines()[0]}")

    # -----------------------------------------------------
    # Classes
    # -----------------------------------------------------

    classes = []

    for name, cls in inspect.getmembers(module, inspect.isclass):

        if cls.__module__ == module_name:

            classes.append((name, cls))

    if classes:

        print("\nClasses")

        for name, cls in classes:

            print()

            print(name)

            for method_name, method in inspect.getmembers(
                cls,
                inspect.isfunction
            ):

                if method.__qualname__.startswith(cls.__name__):

                    print(
                        f"    {method_name}{inspect.signature(method)}"
                    )


# ---------------------------------------------------------
# Scan package
# ---------------------------------------------------------

def summarize_package(package_name):

    package = importlib.import_module(package_name)

    print()
    print("#" * 80)
    print(package_name.upper())
    print("#" * 80)

    summarize_module(package_name)

    if hasattr(package, "__path__"):

        for _, modname, _ in pkgutil.walk_packages(
            package.__path__,
            package.__name__ + "."
        ):

            summarize_module(modname)


# ---------------------------------------------------------
# Entire project
# ---------------------------------------------------------

def summarize_project():

    packages = [

        "analysis",

        "instrument",

        "measurements",

    ]

    for package in packages:

        summarize_package(package)


# ---------------------------------------------------------
# Main
# ---------------------------------------------------------

if __name__ == "__main__":

    summarize_project()
```
